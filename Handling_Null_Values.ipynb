```python
import numpy as np
import pandas as pd

df = pd.read_csv('/content/diabetes_dataset.csv')
```

```python
df.isnull().sum()
```

```python
handling_null = pd.DataFrame()
```

```python
handling_null['bmi'] = df['bmi'].copy()
```

```python
handling_null['insulin_mu_l'] = df['insulin_mu_l'].copy()
```

```python
handling_null
```

```python
for i in handling_null.columns:
  handling_null[i+'mean'] = handling_null[i].fillna(handling_null[i].mean())
  handling_null[i+'median'] = handling_null[i].fillna(handling_null[i].median())
  handling_null[i+'mode'] = handling_null[i].fillna(handling_null[i].mode()[0])
```

```python
handling_null.sample(20)
```

```python
handling_null['bmi_ran'] = handling_null['bmi']
s = handling_null['bmi'].dropna().sample(handling_null['bmi'].isnull().sum())
s.index = handling_null[handling_null['bmi'].isnull()].index
handling_null.loc[handling_null['bmi'].isnull(), 'bmi_ran'] = s
```

```python
handling_null['insulin_mu_l_ran'] = handling_null['insulin_mu_l']
s = handling_null['insulin_mu_l'].dropna().sample(handling_null['insulin_mu_l'].isnull().sum())
s.index = handling_null[handling_null['insulin_mu_l'].isnull()].index
handling_null.loc[handling_null['insulin_mu_l'].isnull(), 'insulin_mu_l_ran'] = s
```
